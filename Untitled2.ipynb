```python

from IPython.display import HTML, display
import uuid

game_id = "snake_" + uuid.uuid4().hex[:8]

html = r"""
<div id="GAME_ID" class="nokia-wrapper">
  <style>
    #GAME_ID {
      font-family: monospace;
      background: #202a24;
      padding: 22px 10px;
      border-radius: 18px;
      text-align: center;
      color: #d9e8d5;
      max-width: 500px;
      margin: auto;
    }

    #GAME_ID .phone {
      background: #b6c8a6;
      border: 8px solid #d1d2c7;
      border-radius: 30px;
      padding: 18px;
      max-width: 380px;
      margin: auto;
      box-shadow: 0 8px 20px #0008;
    }

    #GAME_ID .brand {
      color: #354b38;
      font-size: 25px;
      font-weight: bold;
      letter-spacing: 5px;
      margin-bottom: 12px;
    }

    #GAME_ID .screen {
      background: #9bbc75;
      border: 5px solid #53694c;
      padding: 8px;
      position: relative;
    }

    #GAME_ID .scoreboard {
      display: flex;
      justify-content: space-between;
      color: #243d25;
      font-size: 13px;
      font-weight: bold;
      padding: 3px;
    }

    #GAME_ID canvas {
      width: 100%;
      display: block;
      background: #9bbc75;
      image-rendering: pixelated;
      touch-action: none;
    }

    #GAME_ID .message {
      color: #263d25;
      font-size: 17px;
      font-weight: bold;
      min-height: 25px;
      padding: 5px;
    }

    #GAME_ID .controls {
      margin-top: 15px;
    }

    #GAME_ID button {
      background: #344a3a;
      color: #d9e8d5;
      border: 2px solid #243528;
      border-radius: 7px;
      padding: 10px 15px;
      margin: 4px;
      font-family: monospace;
      font-weight: bold;
      cursor: pointer;
    }

    #GAME_ID button:hover {
      background: #526b4d;
    }

    #GAME_ID .instructions {
      color: #30432f;
      font-size: 12px;
      margin-top: 12px;
    }

    #GAME_ID .dpad {
      margin-top: 15px;
      display: grid;
      grid-template-columns: 48px 48px 48px;
      justify-content: center;
      gap: 4px;
    }

    #GAME_ID .dpad button {
      margin: 0;
      padding: 10px 0;
      font-size: 17px;
    }

    #GAME_ID .empty {
      visibility: hidden;
    }
  </style>

  <div class="phone">
    <div class="brand">NOKIA</div>

    <div class="screen">
      <div class="scoreboard">
        <span>SCORE: <span id="score">0</span></span>
        <span>BEST: <span id="best">0</span></span>
      </div>

      <canvas id="canvas" width="360" height="360"></canvas>

      <div class="message" id="message">
        PRESS START
      </div>
    </div>

    <div class="controls">
      <button id="start">▶ START</button>
      <button id="pause">Ⅱ PAUSE</button>
      <button id="restart">↻ RESTART</button>
    </div>

    <div class="dpad">
      <button class="empty"> </button>
      <button data-dir="up">▲</button>
      <button class="empty"> </button>
      <button data-dir="left">◀</button>
      <button data-dir="down">▼</button>
      <button data-dir="right">▶</button>
    </div>

    <div class="instructions">
      USE ARROW KEYS OR BUTTONS TO MOVE<br>
      EAT FOOD • GROW LONGER • AVOID WALLS
    </div>
  </div>
</div>

<script>
(() => {
  const root = document.getElementById("GAME_ID");
  if (!root || root.dataset.initialized) return;
  root.dataset.initialized = "true";

  const canvas = root.querySelector("#canvas");
  const ctx = canvas.getContext("2d");
  const scoreEl = root.querySelector("#score");
  const bestEl = root.querySelector("#best");
  const message = root.querySelector("#message");

  const SIZE = 18;
  const CELLS = 20;
  const COLORS = {
    background: "#9bbc75",
    dark: "#263d25",
    light: "#b7ce8c"
  };

  let snake, food, direction, nextDirection;
  let score = 0;
  let best = 0;
  let timer = null;
  let running = false;
  let paused = false;
  let gameOver = false;
  let soundEnabled = true;
  let audioContext = null;

  function beep(freq = 500, duration = 0.07) {
    if (!soundEnabled) return;
    try {
      audioContext ||= new (window.AudioContext ||
                            window.webkitAudioContext)();
      if (audioContext.state === "suspended") audioContext.resume();

      const oscillator = audioContext.createOscillator();
      const gain = audioContext.createGain();

      oscillator.type = "square";
      oscillator.frequency.value = freq;
      gain.gain.value = 0.035;

      oscillator.connect(gain);
      gain.connect(audioContext.destination);
      oscillator.start();
      oscillator.stop(audioContext.currentTime + duration);
    } catch (e) {}
  }

  function reset() {
    clearInterval(timer);
    snake = [
      {x: 9, y: 10},
      {x: 8, y: 10},
      {x: 7, y: 10}
    ];

    direction = {x: 1, y: 0};
    nextDirection = {x: 1, y: 0};
    score = 0;
    running = false;
    paused = false;
    gameOver = false;

    placeFood();
    updateScore();
    message.textContent = "PRESS START";
    draw();
  }

  function placeFood() {
    do {
      food = {
        x: Math.floor(Math.random() * CELLS),
        y: Math.floor(Math.random() * CELLS)
      };
    } while (snake.some(part =>
      part.x === food.x && part.y === food.y
    ));
  }

  function updateScore() {
    scoreEl.textContent = score;
    bestEl.textContent = best;
  }

  function draw() {
    ctx.fillStyle = COLORS.background;
    ctx.fillRect(0, 0, canvas.width, canvas.height);

    // Pixel grid
    ctx.strokeStyle = "rgba(38,61,37,0.07)";
    for (let i = 0; i <= CELLS; i++) {
      ctx.beginPath();
      ctx.moveTo(i * SIZE, 0);
      ctx.lineTo(i * SIZE, canvas.height);
      ctx.stroke();

      ctx.beginPath();
      ctx.moveTo(0, i * SIZE);
      ctx.lineTo(canvas.width, i * SIZE);
      ctx.stroke();
    }

    // Food
    ctx.fillStyle = COLORS.dark;
    ctx.fillRect(
      food.x * SIZE + 4,
      food.y * SIZE + 4,
      SIZE - 8,
      SIZE - 8
    );

    // Snake
    snake.forEach((part, i) => {
      ctx.fillStyle = i === 0 ? "#172d1a" : COLORS.dark;
      ctx.fillRect(
        part.x * SIZE + 1,
        part.y * SIZE + 1,
        SIZE - 2,
        SIZE - 2
      );

      if (i === 0) {
        ctx.fillStyle = COLORS.light;
        ctx.fillRect(part.x * SIZE + 5,
                     part.y * SIZE + 5, 3, 3);
        ctx.fillRect(part.x * SIZE + 11,
                     part.y * SIZE + 5, 3, 3);
      }
    });

    if (paused || gameOver) {
      ctx.fillStyle = "rgba(155,188,117,0.72)";
      ctx.fillRect(0, 0, canvas.width, canvas.height);
      ctx.fillStyle = COLORS.dark;
      ctx.textAlign = "center";
      ctx.font = "bold 23px monospace";
      ctx.fillText(
        gameOver ? "GAME OVER" : "PAUSED",
        canvas.width / 2,
        canvas.height / 2
      );
    }
  }

  function setDirection(dir) {
    if (gameOver) return;

    const dirs = {
      up:    {x: 0, y: -1},
      down:  {x: 0, y: 1},
      left:  {x: -1, y: 0},
      right: {x: 1, y: 0}
    };

    const proposed = dirs[dir];
    if (!proposed) return;

    if (
      proposed.x === -direction.x &&
      proposed.y === -direction.y
    ) return;

    nextDirection = proposed;

    if (!running && !paused) startGame();
  }

  function tick() {
    direction = nextDirection;

    const head = {
      x: snake[0].x + direction.x,
      y: snake[0].y + direction.y
    };

    const eating =
      head.x === food.x && head.y === food.y;

    const body = eating ? snake : snake.slice(0, -1);

    if (
      head.x < 0 || head.x >= CELLS ||
      head.y < 0 || head.y >= CELLS ||
      body.some(part => part.x === head.x && part.y === head.y)
    ) {
      endGame();
      return;
    }

    snake.unshift(head);

    if (eating) {
      score += 10;
      best = Math.max(best, score);
      updateScore();
      placeFood();
      beep(750, 0.09);

      clearInterval(timer);
      timer = setInterval(tick, Math.max(65, 145 - score));
    } else {
      snake.pop();
    }

    draw();
  }

  function startGame() {
    if (gameOver || running) return;

    running = true;
    paused = false;
    message.textContent = "PLAYING";
    clearInterval(timer);
    timer = setInterval(tick, Math.max(65, 145 - score));
    draw();
    beep(500);
  }

  function pauseGame() {
    if (gameOver) return;

    if (!running && !paused) return;

    if (running) {
      clearInterval(timer);
      running = false;
      paused = true;
      message.textContent = "PAUSED";
    } else {
      paused = false;
      startGame();
    }

    draw();
  }

  function endGame() {
    clearInterval(timer);
    running = false;
    paused = false;
    gameOver = true;
    message.textContent = "GAME OVER! SCORE: " + score;
    beep(180, 0.35);
    draw();
  }

  function restartGame() {
    reset();
    startGame();
  }

  document.addEventListener("keydown", e => {
    const keys = {
      ArrowUp: "up",
      ArrowDown: "down",
      ArrowLeft: "left",
      ArrowRight: "right"
    };

    if (keys[e.key]) {
      e.preventDefault();
      setDirection(keys[e.key]);
    }

    if (e.key === " " || e.key.toLowerCase() === "p") {
      e.preventDefault();
      pauseGame();
    }
  });

  root.querySelectorAll("[data-dir]").forEach(button => {
    button.addEventListener("click", () =>
      setDirection(button.dataset.dir)
    );
  });

  root.querySelector("#start").addEventListener("click", startGame);
  root.querySelector("#pause").addEventListener("click", pauseGame);
  root.querySelector("#restart").addEventListener("click", restartGame);

  reset();
})();
</script>
""".replace("GAME_ID", game_id)

display(HTML(html))
```
