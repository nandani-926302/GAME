```python

from IPython.display import HTML, display

html = r'''
<div id="neonGame" style="
background:#050816;
padding:15px;
border-radius:20px;
max-width:850px;
margin:auto;
color:white;
font-family:Arial;
text-align:center;
border:2px solid #00eaff;
box-shadow:0 0 25px #00d9ff55;
">

<h1 style="color:#00f5ff;text-shadow:0 0 15px #00eaff">
🐍 NEON SNAKE
</h1>

<p style="color:#bd9cff">THE ULTIMATE NEON ARCADE GAME</p>

<div style="display:flex;justify-content:space-around;flex-wrap:wrap;gap:12px">
<div>🏆 SCORE: <b id="score">0</b></div>
<div>🥇 BEST: <b id="best">0</b></div>
<div>⚡ LEVEL: <b id="level">1</b></div>
</div>

<div style="position:relative;margin:15px auto;max-width:800px">

<canvas id="snakeCanvas"
width="800" height="600"
tabindex="0"
style="
width:100%;
height:auto;
background:#070d20;
border:2px solid #174b75;
border-radius:12px;
display:block;
outline:none;
">
</canvas>

<div id="overlay" style="
position:absolute;
inset:0;
background:#030817eF;
display:flex;
flex-direction:column;
align-items:center;
justify-content:center;
border-radius:12px;
">

<h1 id="overlayTitle" style="
font-size:40px;
color:#00f5ff;
text-shadow:0 0 20px #00eaff;
">NEON SNAKE</h1>

<p id="overlayMessage">Ready for an adventure?</p>

<button id="startBtn" style="
background:#00eaff;
color:#00101b;
padding:14px 38px;
border:0;
border-radius:30px;
font-weight:bold;
font-size:18px;
cursor:pointer;
">
START GAME
</button>
</div>
</div>

<div style="display:flex;justify-content:center;gap:10px;flex-wrap:wrap">

<button class="ngbtn" id="pauseBtn">⏸ PAUSE</button>
<button class="ngbtn" id="restartBtn">🔄 RESTART</button>
<button class="ngbtn" id="soundBtn">🔊 SOUND</button>

</div>

<h3 style="color:#00eaff">🎮 GAME CONTROLS</h3>

<div style="display:grid;grid-template-columns:repeat(3,65px);
justify-content:center;gap:7px;margin:15px">

<div></div>
<button class="ngbtn arrow" data-dir="up">⬆️</button>
<div></div>

<button class="ngbtn arrow" data-dir="left">⬅️</button>
<button class="ngbtn arrow" data-dir="down">⬇️</button>
<button class="ngbtn arrow" data-dir="right">➡️</button>

</div>

<p style="color:#9bb5d5;font-size:13px">
Keyboard: Arrow Keys / W A S D | P = Pause
</p>

<h3 style="color:#bd9cff">⚡ SELECT DIFFICULTY</h3>

<div style="display:flex;justify-content:center;gap:8px;flex-wrap:wrap">
<button class="ngbtn diff" data-level="0">EASY</button>
<button class="ngbtn diff" data-level="1">MEDIUM</button>
<button class="ngbtn diff" data-level="2">HARD</button>
<button class="ngbtn diff" data-level="3">EXTREME</button>
</div>

<style>
#neonGame .ngbtn {
background:#101c35;
color:#00eaff;
border:1px solid #00eaff;
border-radius:10px;
padding:11px 16px;
cursor:pointer;
font-weight:bold;
}
#neonGame .ngbtn:hover {
background:#123b59;
box-shadow:0 0 12px #00eaff;
}
#neonGame .arrow {
font-size:23px;
width:65px;
height:52px;
padding:4px;
touch-action:manipulation;
}
#neonGame button:active {
transform:scale(.94);
}
</style>

</div>

<script>
(() => {

const root=document.getElementById("neonGame");
const canvas=root.querySelector("#snakeCanvas");
const ctx=canvas.getContext("2d");

const $=id=>root.querySelector("#"+id);

const W=800,H=600,CELL=25;
const COLS=32,ROWS=24;

let snake=[];
let direction={x:1,y:0};
let nextDirection={x:1,y:0};
let food=null,foodType="normal";

let score=0,best=0,level=1;
let running=false,paused=false,gameOver=false;
let particles=[],stars=[];
let difficulty=1,frame=0,lastMove=0;
let soundOn=true,audio=null;
let animationId=null;
let active=true;

const speeds=[160,120,90,60];

try {
best=Number(localStorage.getItem("neonSnakeBest"))||0;
}catch(e){}

$("best").textContent=best;

for(let i=0;i<100;i++){
stars.push({
x:Math.random()*W,
y:Math.random()*H,
speed:.3+Math.random(),
size:Math.random()*2+1
});
}

function beep(freq=600,duration=.1){
if(!soundOn)return;

try{
audio=audio||new(window.AudioContext||window.webkitAudioContext)();

if(audio.state==="suspended")audio.resume();

let osc=audio.createOscillator();
let gain=audio.createGain();

osc.connect(gain);
gain.connect(audio.destination);

osc.frequency.value=freq;
gain.gain.value=.06;

osc.start();

gain.gain.exponentialRampToValueAtTime(
.001,audio.currentTime+duration
);

osc.stop(audio.currentTime+duration);

}catch(e){}
}

function updateHUD(){
$("score").textContent=score;
$("best").textContent=best;
$("level").textContent=level;
}

function spawnFood(){

let free=[];

for(let x=0;x<COLS;x++){
for(let y=0;y<ROWS;y++){

if(!snake.some(s=>s.x===x&&s.y===y)){
free.push({x,y});
}

}}

if(free.length===0){
winGame();
return;
}

food=free[Math.floor(Math.random()*free.length)];

let chance=Math.random();

foodType=chance<.12?"gold":
chance<.23?"power":"normal";

}

function reset(){

snake=[
{x:10,y:12},
{x:9,y:12},
{x:8,y:12},
{x:7,y:12}
];

direction={x:1,y:0};
nextDirection={x:1,y:0};

score=0;
level=1;
particles=[];

running=true;
paused=false;
gameOver=false;

spawnFood();
updateHUD();

$("overlay").style.display="none";
$("pauseBtn").textContent="⏸ PAUSE";

lastMove=performance.now();

canvas.focus();

}

function setDirection(x,y){

if(x===-direction.x&&y===-direction.y)return;

nextDirection={x,y};

}

function burst(x,y,color){

for(let i=0;i<28;i++){

let angle=Math.random()*Math.PI*2;
let speed=Math.random()*4+1;

particles.push({
x:x*CELL+CELL/2,
y:y*CELL+CELL/2,
vx:Math.cos(angle)*speed,
vy:Math.sin(angle)*speed,
life:40,
color
});

}

}

function endGame(){

running=false;
gameOver=true;

beep(160,.5);

$("overlayTitle").textContent="GAME OVER!";
$("overlayTitle").style.color="#ff399b";

$("overlayMessage").textContent=
"YOUR FINAL SCORE: "+score;

$("startBtn").textContent="PLAY AGAIN";
$("overlay").style.display="flex";

}

function winGame(){

running=false;
gameOver=true;

$("overlayTitle").textContent="YOU WIN!";
$("overlayTitle").style.color="#ffd700";

$("overlayMessage").textContent=
"AMAZING! YOU FILLED THE BOARD!";

$("startBtn").textContent="PLAY AGAIN";
$("overlay").style.display="flex";

}

function move(){

direction=nextDirection;

let head={
x:snake[0].x+direction.x,
y:snake[0].y+direction.y
};

let eating=head.x===food.x&&head.y===food.y;

let body=eating?snake:snake.slice(0,-1);

if(
head.x<0||head.x>=COLS||
head.y<0||head.y>=ROWS||
body.some(s=>s.x===head.x&&s.y===head.y)
){
endGame();
return;
}

snake.unshift(head);

if(eating){

let color="#00ff88";

if(foodType==="gold"){
score+=50;
color="#ffd700";
beep(1000,.2);
}
else if(foodType==="power"){
score+=25;
color="#a855ff";
beep(850,.15);
}
else{
score+=10;
color="#ff40b5";
beep(650,.1);
}

burst(head.x,head.y,color);

level=Math.min(10,1+Math.floor(score/100));

if(score>best){
best=score;
try{
localStorage.setItem("neonSnakeBest",best);
}catch(e){}
}

spawnFood();

}else{
snake.pop();
}

updateHUD();

}

function drawBackground(){

let gradient=ctx.createLinearGradient(0,0,W,H);

gradient.addColorStop(0,"#071328");
gradient.addColorStop(1,"#100725");

ctx.fillStyle=gradient;
ctx.fillRect(0,0,W,H);

stars.forEach(s=>{

s.y+=s.speed;

if(s.y>H){
s.y=0;
s.x=Math.random()*W;
}

ctx.fillStyle="rgba(100,190,255,.65)";
ctx.fillRect(s.x,s.y,s.size,s.size);

});

ctx.strokeStyle="rgba(40,100,150,.13)";
ctx.lineWidth=1;

for(let x=0;x<=W;x+=CELL){
ctx.beginPath();
ctx.moveTo(x,0);
ctx.lineTo(x,H);
ctx.stroke();
}

for(let y=0;y<=H;y+=CELL){
ctx.beginPath();
ctx.moveTo(0,y);
ctx.lineTo(W,y);
ctx.stroke();
}

}

function drawFood(){

if(!food)return;

let x=food.x*CELL+CELL/2;
let y=food.y*CELL+CELL/2;

let pulse=1+Math.sin(frame*.13)*.15;

let color=foodType==="gold"?"#ffd700":
foodType==="power"?"#b35cff":"#ff36a7";

ctx.save();

ctx.shadowBlur=28;
ctx.shadowColor=color;
ctx.fillStyle=color;

ctx.beginPath();
ctx.arc(x,y,9*pulse,0,Math.PI*2);
ctx.fill();

ctx.shadowBlur=0;
ctx.fillStyle="white";

ctx.beginPath();
ctx.arc(x-3,y-3,2,0,Math.PI*2);
ctx.fill();

ctx.restore();

}

function drawSnake(){

snake.forEach((s,i)=>{

let x=s.x*CELL+CELL/2;
let y=s.y*CELL+CELL/2;

let hue=(frame*1.5+i*12)%360;

let color=i===0?"#00f5ff":
`hsl(${hue},100%,60%)`;

ctx.save();

ctx.shadowBlur=i===0?25:15;
ctx.shadowColor=color;
ctx.fillStyle=color;

let size=i===0?11:10;

ctx.beginPath();
ctx.roundRect(
x-size,y-size,size*2,size*2,6
);
ctx.fill();

ctx.restore();

if(i===0){

let dx=direction.x;
let dy=direction.y;

for(let side of [-1,1]){

let ex=x+dx*5-dy*side*5;
let ey=y+dy*5+dx*side*5;

ctx.fillStyle="white";
ctx.beginPath();
ctx.arc(ex,ey,3.5,0,Math.PI*2);
ctx.fill();

ctx.fillStyle="#071328";
ctx.beginPath();
ctx.arc(ex+dx,ey+dy,1.7,0,Math.PI*2);
ctx.fill();

}

}

});

}

function drawParticles(){

particles=particles.filter(p=>p.life>0);

particles.forEach(p=>{

p.x+=p.vx;
p.y+=p.vy;
p.life--;

ctx.globalAlpha=Math.max(0,p.life/40);
ctx.fillStyle=p.color;

ctx.beginPath();
ctx.arc(p.x,p.y,3,0,Math.PI*2);
ctx.fill();

});

ctx.globalAlpha=1;

}

function draw(){

frame++;

drawBackground();
drawFood();
drawSnake();
drawParticles();

if(paused&&!gameOver){

ctx.fillStyle="rgba(0,0,0,.55)";
ctx.fillRect(0,0,W,H);

ctx.fillStyle="#00f5ff";
ctx.font="bold 55px Arial";
ctx.textAlign="center";
ctx.shadowBlur=20;
ctx.shadowColor="#00f5ff";

ctx.fillText("PAUSED",W/2,H/2);

ctx.shadowBlur=0;

}

}

function loop(timestamp){

if(!active)return;

if(running&&!paused&&!gameOver){

let interval=Math.max(
45,
speeds[difficulty]-(level-1)*5
);

if(timestamp-lastMove>=interval){

move();
lastMove=timestamp;

}

}

draw();

animationId=requestAnimationFrame(loop);

}

// Keyboard controls

function keyHandler(e){

const keys=[
"ArrowUp","ArrowDown","ArrowLeft","ArrowRight",
"w","a","s","d"
];

if(keys.includes(e.key))e.preventDefault();

if(e.key==="ArrowUp"||e.key.toLowerCase()==="w")
setDirection(0,-1);

if(e.key==="ArrowDown"||e.key.toLowerCase()==="s")
setDirection(0,1);

if(e.key==="ArrowLeft"||e.key.toLowerCase()==="a")
setDirection(-1,0);

if(e.key==="ArrowRight"||e.key.toLowerCase()==="d")
setDirection(1,0);

if(e.key.toLowerCase()==="p"&&running&&!gameOver){

paused=!paused;

$("pauseBtn").textContent=
paused?"▶ RESUME":"⏸ PAUSE";

lastMove=performance.now();

}

}

document.addEventListener("keydown",keyHandler);

// Screen controls

root.querySelectorAll(".arrow").forEach(btn=>{

btn.addEventListener("click",()=>{

let d=btn.dataset.dir;

if(d==="up")setDirection(0,-1);
if(d==="down")setDirection(0,1);
if(d==="left")setDirection(-1,0);
if(d==="right")setDirection(1,0);

});

});

// Start

$("startBtn").onclick=()=>{
reset();
beep(500);
};

$("restartBtn").onclick=()=>{
reset();
};

$("pauseBtn").onclick=()=>{

if(!running||gameOver)return;

paused=!paused;

$("pauseBtn").textContent=
paused?"▶ RESUME":"⏸ PAUSE";

lastMove=performance.now();

};

$("soundBtn").onclick=()=>{

soundOn=!soundOn;

$("soundBtn").textContent=
soundOn?"🔊 SOUND":"🔇 MUTED";

if(soundOn)beep();

};

root.querySelectorAll(".diff").forEach(btn=>{

btn.addEventListener("click",()=>{

difficulty=Number(btn.dataset.level);

root.querySelectorAll(".diff").forEach(b=>{
b.style.background="#101c35";
});

btn.style.background="#15516b";

});

});

// Start with a visible snake behind the welcome overlay
reset();
running=false;

draw();

animationId=requestAnimationFrame(loop);

})();
</script>
'''

display(HTML(html))
```
